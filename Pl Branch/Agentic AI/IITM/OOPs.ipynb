# Single level Inheritance

```python
# Class A ----> Class B

# Parent class ----> Child class
```

```python
class Parent:
    def __init__(self):
        self.name = 'john'

class child(Parent):
    def display(self):
        print(self.name)

obj = child()
obj.display()
```

# Multi level inheritance

```python
Class A ---- > Class B ----> Class C
```

```python
class Parent:
    def __init__(self):
        self.name = "Rohan"

class Child(Parent):
    def __init__(self):
        Parent.__init__(self)
        self.age = 40

class Child2(Child):
    def __init__(self):
        Child.__init__(self)
        self.prof = 'banker'

    def display(self):
        print(self.name,self.age,self.prof)

obj = Child2()
obj.display()
```

```python
class Parent:
    def __init__(self):
        self.name = "Shampoo"


class Child1(Parent):
    def __init__(self):
        Parent.__init__(self)
        self.weight = "500 Ml"


class Child2(Child1):
    def __init__(self):
        Child1.__init__(self)
        self.price = "500 Rupees"


class Child3(Child2):
    def __init__(self):
        Child2.__init__(self)
        self.available_quantity = "5 Pieces"
    def display(self):
        print(self.name, self.weight, self.price, self.available_quantity)


obj = Child3()
obj.display()
```

```python
class ProductName:
    def __init__(self):
        self.name = 'Tata'

class ProductWeight(ProductName):
    def __init__(self):
        ProductName.__init__(self)
        self.Weight = '55'


class ProductPricing(ProductWeight):
    def __init__(self):
        ProductWeight.__init__(self)
        self.Price = '1000'


class ProductQuantity(ProductPricing):
    def __init__(self):
        ProductPricing.__init__(self)
        self.Qty = '1'


    def display(self):
        print(self.name, self.Weight, self.Price, self.Qty)


obj = ProductQuantity()
obj.display()
```

```python
class Product:
  def __init__(self,name):
    self.name = name


class ProductPrice(Product):
    def __init__(self,name,price):
      Product.__init__(self,name)
      self.price = price


class ProductWeight(ProductPrice):

    def __init__(self,name,price,weight):
      ProductPrice.__init__(self,name,price)
      self.weight = weight


class ProductAvailibilty(ProductWeight):

    def __init__(self,name,price,weight,available):
      Product.__init__(self,name,price,weight)
      self.available = available



    def display(self):
      print(self.name,self.price,self.weight,self.available)


ob=ProductAvailibilty("yash","2","2","y")
ob.display()
```
